```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
from PIL import Image
```

```python
img_pil = Image.open("sample_color.jpg")
img_pil_arr = np.array(img_pil)
```

```python
print(img_pil_arr.shape)
```

```python
print(img_pil_arr.dtype)
```

```python
plt.imshow(img_pil_arr)
plt.axis("off")
plt.show()
```

```python
img_cv = cv2.imread("sample_color.jpg")
```

```python
print(img_cv.shape)
print(img_cv.dtype)
```

```python
plt.imshow(img_cv)
plt.show()
```

```python
img_cv_rgb = cv2.cvtColor(img_cv, cv2.COLOR_BGR2RGB)
plt.imshow(img_cv_rgb)
plt.show()
```

```python
img_gray = cv2.imread("sample_gray.jpg", cv2.IMREAD_GRAYSCALE)
print(img_gray.shape)
```

```python
plt.imshow(img_gray, cmap='gray')
plt.axis("off")
plt.show()
```

```python
forced_gray = cv2.imread("sample_color.jpg", cv2.IMREAD_GRAYSCALE)
plt.imshow(forced_gray, cmap='gray')
plt.axis('off')
plt.show()
```

```python
cv2.imwrite('color_to_gray.jpg', forced_gray)
```
