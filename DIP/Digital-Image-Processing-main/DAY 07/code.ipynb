```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

1. Load grayscale image directly using openCV 

```python
img = cv2.imread("image1.png", cv2.IMREAD_GRAYSCALE).astype(np.float32)
img_uint8 = cv2.imread("image1.png", cv2.IMREAD_GRAYSCALE)
```

```python
plt.figure(figsize=(6,5))
plt.imshow(img, cmap='gray', vmin=0, vmax=255)
plt.title('Original image')
plt.axis('off')
plt.show()
```

2. Brightness adjustment: add or substract a constant

```python
brighter = img + 60
brighter = np.clip(brighter, 0, 255)
brighter = brighter.astype(np.uint8)
```

```python
darker = img - 60
darker = np.clip(darker, 0, 255)
darker = darker.astype(np.uint8)
```

```python
plt.figure(figsize=(10,5))
plt.subplot(1,3,1)
plt.imshow(darker, cmap='gray', vmin=0, vmax=255)
plt.title("Darker (-60)")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(img.astype(np.uint8), cmap='gray', vmin=0, vmax=255)
plt.title("Original image")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(brighter, cmap='gray', vmin=0, vmax=255)
plt.title("Brighter (+60)")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
brightness_matrix = np.full(img_uint8.shape, 60, dtype=np.int16)
darker2 = np.clip(img_uint8.astype(np.int16) - brightness_matrix, 0, 255).astype(np.uint8)
brighter2 = np.clip(img_uint8.astype(np.int16) + brightness_matrix, 0, 255).astype(np.uint8)

# cv2.imshow does not work in Jupyter notebooks, so use matplotlib instead
plt.figure(figsize=(15,4))
plt.subplot(1,3,1)
plt.imshow(darker2, cmap="gray", vmin=0, vmax=255)
plt.title("Darker (-60)")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(img_uint8, cmap="gray", vmin=0, vmax=255)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(brighter2, cmap="gray", vmin=0, vmax=255)
plt.title("Brighter (+60)")
plt.axis("off")

plt.tight_layout()
plt.show()
```

3. Contrast Stretching

```python
min_val = np.min(img)
max_val = np.max(img)

# Manual contrast stretching
stretched = (img - min_val) * (255.0 / (max_val - min_val))
stretched = np.clip(stretched, 0, 255).astype(np.uint8)
```

```python
print("Before stretching -> min:", min_val, "max:", max_val)
print("After stretching -> min:", stretched.min(), "max:", stretched.max())
```

```python
stretched_cv2 = cv2.normalize(img_uint8, None, 0, 255, cv2.NORM_MINMAX)
```

```python
plt.figure(figsize=(10,5))
plt.subplot(1,3,1)
plt.imshow(img.astype(np.uint8), cmap="gray", vmin=0, vmax=255)
plt.title("Original (low contrast)")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(stretched, cmap="gray", vmin=0, vmax=255)
plt.title("manual contrast stretch")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(stretched_cv2, cmap="gray", vmin=0, vmax=255)
plt.title("cv2.normalize stretch")
plt.axis("off")

plt.tight_layout()
plt.show()
```

4. Image negative/ inversion

```python
negative = 255 - img.astype(np.uint8)
negative_cv2 = cv2.bitwise_not(img_uint8)
```

```python
plt.figure(figsize=(10,5))
plt.subplot(1,3,1)
plt.imshow(img.astype(np.uint8), cmap="gray", vmin=0, vmax=255)
plt.title("Original")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(negative, cmap="gray", vmin=0, vmax=255)
plt.title("Manual Negative (255-px)")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(negative_cv2, cmap="gray", vmin=0, vmax=255)
plt.title("cv2 bitwise_not negative")
plt.axis("off")

plt.tight_layout()
plt.show()
```

5. Threshold binarization using openCV

```python
_, bin_img_80 = cv2.threshold(img_uint8, 80, 255, cv2.THRESH_BINARY)
```

```python
_, bin_img_128 = cv2.threshold(img_uint8, 128, 255, cv2.THRESH_BINARY)
```

```python
_, bin_img_180 = cv2.threshold(img_uint8, 180, 255, cv2.THRESH_BINARY)
```

```python
plt.figure(figsize=(10,5))
plt.subplot(1,3,1)
plt.imshow(bin_img_80, cmap="gray", vmin=0, vmax=255)
plt.title("bin_img_80")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(bin_img_128, cmap="gray", vmin=0, vmax=255)
plt.title("bin_img_128")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(bin_img_180, cmap="gray", vmin=0, vmax=255)
plt.title("bin_img_180")
plt.axis("off")

plt.tight_layout()
plt.show()
```
