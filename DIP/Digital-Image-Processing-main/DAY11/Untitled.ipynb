```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
img=cv2.imread("sample_image.jpg",cv2.IMREAD_GRAYSCALE)
```

```python
plt.figure(figsize=(6,5))
plt.imshow(img,cmap="gray")
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
laplacian_kernal=np.array([[0,1,0],
                          [1,-4,1],
                          [0,1,0]],dtype=np.float32)
```

```python
laplacian_manual=cv2.filter2D(img.astype(np.float32),-1,laplacian_kernal)
laplacian_manual_display=cv2.convertScaleAbs(laplacian_manual)
```

```python
laplacian_cv2=cv2.Laplacian(img,cv2.CV_64F,ksize=3)
laplacian_cv2_display=(cv2.convertScaleAbs(laplacian_cv2))
```

```python
plt.figure(figsize=(16,5))
plt.subplot(1,3,1)
plt.imshow(img,cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(laplacian_manual_display,cmap="gray")
plt.title("laplacian_manual_display")
plt.axis("off")


plt.subplot(1,3,3)
plt.imshow(laplacian_cv2_display,cmap="gray")
plt.title("laplacian_cv2_display")
plt.axis("off")


plt.show()
```

```python
sobel_kernel_x=np.array([[-1,0,1],
                        [-2,0,2],
                        [-1,0,1]],dtype=np.float32)

```

```python
sobel_kernel_y=np.array([[-1,-2,-1],
                        [0,0,0],
                        [1,2,1]],dtype=np.float32)

```

```python
sobel_x_manual=cv2.filter2D(img.astype(np.float32),-1,sobel_kernel_x)
sobel_y_manual=cv2.filter2D(img.astype(np.float32),-1,sobel_kernel_y)
sobel_manual_magnitute=cv2.magnitude(sobel_x_manual,sobel_y_manual)
sobel_manual_display=cv2.convertScaleAbs(sobel_manual_magnitute)
```

```python
plt.figure(figsize=(16,5))
plt.subplot(1,4,1)
plt.imshow(img,cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,4,2)
plt.imshow(sobel_x_manual,cmap="gray")
plt.title("sobel_x_manual")
plt.axis("off")


plt.subplot(1,4,3)
plt.imshow(sobel_y_manual,cmap="gray")
plt.title("sobel_y_manual")
plt.axis("off")



plt.subplot(1,4,4)
plt.imshow(sobel_manual_display,cmap="gray")
plt.title("sobel_manual_display")
plt.axis("off")

plt.show()
```

## Sobel filter,OpenCv's dedicated built-in cv2.Sobel function

```python
sobel_x_cv2=cv2.Sobel(img,cv2.CV_64F,1,0,ksize=3)
sobel_y_cv2=cv2.Sobel(img,cv2.CV_64F,0,1,ksize=3)
sobel_cv2_magnitute=cv2.magnitude(sobel_x_cv2,sobel_y_cv2)
sobel_cv2_display=cv2.convertScaleAbs(sobel_cv2_magnitute)

```

```python
plt.figure(figsize=(16,5))
plt.subplot(1,4,1)
plt.imshow(img,cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,4,2)
plt.imshow(sobel_x_cv2,cmap="gray")
plt.title("sobel_x_cv2")
plt.axis("off")


plt.subplot(1,4,3)
plt.imshow(sobel_y_cv2,cmap="gray")
plt.title("sobel_y_cv2")
plt.axis("off")



plt.subplot(1,4,4)
plt.imshow(sobel_cv2_display,cmap="gray")
plt.title("sobel_cv2_display")
plt.axis("off")

plt.show()
```

```python
Sharpening_kernal=np.array([[0,-1,0,],
                            [-1,5,-1],
                            [0,-1,0]],dtype=np.float32)
```

```python
sharpened_manual=cv2.filter2D(img,-1,Sharpening_kernal)
```

```python
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img,cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(1,2,2)
plt.title("sharpened_manual")
plt.imshow(sharpened_manual,cmap="gray")
plt.axis("off")



plt.show()
```
