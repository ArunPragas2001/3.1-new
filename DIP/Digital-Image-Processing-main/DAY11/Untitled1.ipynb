```python
import numpy as np
import cv2
import matplotlib.pyplot as plt 

```

```python
img=cv2.imread("images.jpg")
```

```python
laplacian_kernal=np.array([[0,1,0,],
                            [1,-4,1],
                            [0,1,0]],dtype=np.float32)
```

```python
laplacian_manual=cv2.filter2D(img.astype(np.float32),-1,laplacian_kernal)
laplacian_manual_display=cv2.convertScaleAbs(laplacian_manual)
```

```python
plt.figure(figsize=(16,5))
plt.subplot(1,2,1)
plt.imshow(img,cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(laplacian_manual_display,cmap="gray")
plt.title("laplacian_manual_display")
plt.axis("off")
```
