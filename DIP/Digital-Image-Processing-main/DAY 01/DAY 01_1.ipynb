```python
# lists

data = [10, 20, 30, 400, 500, 600]
print("Length is",len(data))

print("First element is",data[0])
print("Last element is",data[-1])

print(data[1:4])
print("Max is",max(data))
print("Min is",min(data))
print("Summation is",sum(data))

data.append(55)
data.insert(3,40)
data.pop()
data.sort()
data.sort(reverse=True)

print(data)
data.clear()
```

```python
# tuples and dictionaries
data1 = (12, 24)
print(data1[0])

data2 = {"width": 256, "height": 256, "channels":3}
print(data2["width"])
print(data2["height"])
```

```python
# conditionals

pixelValue = 130

if pixelValue > 200: print("Bright")
elif pixelValue > 100: print("Medium")
else: print("Dark")
```

```python
# functions

def isPalindrome(text):
    print("Yes" if text == text[::-1] else "No")
    
isPalindrome("hello")
isPalindrome('radar')
```
