```python
pixels = [10, 90, 130, 200, 255, 5]
print(pixels[0])
print(pixels[-1])
print(pixels[2:4])
```

```python
print(max(pixels))
print(min(pixels))
print(sum(pixels) / len(pixels))
```

```python
varPixel = 75
if varPixel > 150: print("Bright pixel")
elif varPixel > 50: print("Medium pixel")
else: print("Dark pixel")
```

```python
for pixel in pixels:
    if pixel > 150:
        print(pixel, ": Bright pixel")
    elif pixel > 50:
        print(pixel, ": Medium pixel")
    else:
        print(pixel, ": Dark pixel")
```

```python
count = 0
for pixel in pixels:
    if pixel > 128:
        count += 1
print(count)
```

```python
def is_dark(pixel):
    return True if pixel < 50 else False
print(is_dark(10))
print(is_dark(60))
print(is_dark(120))
```

```python
def classify_pixel(pixel):
    if pixel > 150:
        return "Bright"
    elif pixel > 50:
        return "Medium"
    else:
        return "Dark"
    
print(classify_pixel(10))
print(classify_pixel(60))
print(classify_pixel(120))
```
